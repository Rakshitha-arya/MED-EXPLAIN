```python
import os
import xml.etree.ElementTree as ET
import pandas as pd

path = "../data/MedQuAD-master"

data = []

for root, dirs, files in os.walk(path):
    for file in files:
        if file.endswith(".xml"):
            file_path = os.path.join(root, file)

            try:
                tree = ET.parse(file_path)
                root_xml = tree.getroot()

                for qa in root_xml.iter("QAPair"):
                    question = qa.findtext("Question")
                    answer = qa.findtext("Answer")

                    if question and answer:
                        data.append({
                            "question": question.strip(),
                            "answer": answer.strip()
                        })

            except Exception:
                pass

df = pd.DataFrame(data)

print("Total QA pairs:", len(df))
print(df.head())
```

```python
# Save the extracted MedQuAD data

output_path = "../data/processed_medquad.csv"

df.to_csv(output_path, index=False)

print("Saved successfully!")
print("File:", output_path)
print("Rows:", len(df))
```
